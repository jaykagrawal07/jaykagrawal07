```python
import random

def play_hangman():
    # 1. Use a small list of 5 predefined words
    word_list = ["python", "algorithm", "variable", "software", "network"]

    # Select a random word from the list and convert to uppercase
    secret_word = random.choice(word_list).upper()

    guessed_letters = []
    incorrect_guesses = 0
    max_incorrect_guesses = 6  # 2. Limit incorrect guesses to 6

    print("Welcome to Hangman!")
    print(f"You can make up to {max_incorrect_guesses} incorrect guesses.")

    # 3. while loop for the main game structure
    while incorrect_guesses < max_incorrect_guesses:
        # Build the current state of the word
        display_word = ""
        for letter in secret_word:
            if letter in guessed_letters:
                display_word += letter + " "
            else:
                display_word += "_ "

        print(f"\nWord: {display_word}")
        print(f"Incorrect guesses left: {max_incorrect_guesses - incorrect_guesses}")

        # Check for win condition
        if "_" not in display_word:
            print(f"\nCongratulations! You guessed the word: {secret_word}")
            break

        # Basic console input/output
        guess = input("Guess a letter: ").upper()

        # 4. if-else statements for game logic
        if len(guess) != 1 or not guess.isalpha():
            print("Invalid input. Please enter a single letter.")
            continue

        if guess in guessed_letters:
            print("You already guessed that letter. Try again!")
            continue

        # Add the valid guess to our list
        guessed_letters.append(guess)

        # Check if the guess is correct
        if guess in secret_word:
            print("Correct!")
        else:
            print("Incorrect!")
            incorrect_guesses += 1

    # Check for loss condition
    if incorrect_guesses == max_incorrect_guesses:
        print(f"\nGame Over! You ran out of guesses.")
        print(f"The secret word was: {secret_word}")

if __name__ == "__main__":
    play_hangman()
```
